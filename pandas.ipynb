## Aggregate Statistics (groupby)

```python
df.loc[df['Type 1'].str.contains(''), ['Genertion', 'Legendary']]=['1',True]
```

```python
df.groupby(['Type 1']).sum(numeric_only=True)
# can use mean, sum, count
```

```python
df.groupby(['Type 1']).mean(numeric_only=True).sort_values('Defense', ascending=False)
```

```python
df['Count']=1
df.groupby(['Type 1']).count()
```

```python
df.groupby(['Type 1']).count()['Count']
```

```python
df.groupby(['Type 1','Type 2']).count()['Count']
```

## Working with large data set

```python
# if data in gbs like 20-30gbs, load it in chunks
for df in pd.read_csv('Modified.csv', chunksize=10):
    print('Chunk Df')
    print(df)
```

```python
new_df=pd.DataFrame(columns=df.columns)

for df in pd.read_csv('Modified.csv',chunksize=10):
    result=df.groupby(['Type 1']).count()
    new_df=pd.concat([result, new_df])
new_df
```
